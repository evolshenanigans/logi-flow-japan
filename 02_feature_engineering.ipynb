```python
import pandas as pd
import jpholiday
from datetime import datetime

df = pd.read_csv('data/synthetic_logistics_data.csv')
df['Date'] = pd.to_datetime(df['Date'])

# Create "Is_Holiday" Feature
# jpholiday.is_holiday(date) returns True/False
def check_holiday(row):
    return jpholiday.is_holiday(row['Date'])

df['is_holiday'] = df.apply(check_holiday, axis=1)
df['weekday'] = df['Date'].dt.weekday

print(df[df['is_holiday'] == True].head())
```

```python
# 5. Encode Categorical Data
# We convert columns to 'category' types, then extract the numeric codes.
categorical_cols = ['Origin_Prefecture', 'Destination_Prefecture', 'Vehicle_Type', 'Status']

for col in categorical_cols:
    df[col] = df[col].astype('category')
    df[f'{col}_Code'] = df[col].cat.codes

# 6. Clean Up
features_df = df.drop(columns=categorical_cols)

print("✅ Encoding Complete.")
print(features_df.head())
print(features_df.info())
```

```python
features_df.to_parquet('data/features_engineered.parquet', index=False)

print("💾 Data saved to 'data/features_engineered.parquet'. Ready for Module 03.")
```
