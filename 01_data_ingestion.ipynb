```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Japanese Context: Major Logistics Hubs
prefectures = ['東京都', '大阪府', '愛知県', '福岡県', '北海道', '宮城県', '広島県']
vehicle_types = ['Large_Truck', 'Medium_Truck', 'Refrigerated', 'Light_Van']
statuses = ['Completed', 'Delayed', 'On_Time']

def generate_logi_data(num_rows=1000):
    data = []
    start_date = datetime(2023, 1, 1)
    
    for _ in range(num_rows):
        # 1. Random Date (up to today)
        days_offset = random.randint(0, 400)
        current_date = start_date + timedelta(days=days_offset)
        
        # 2. Origin & Destination (Ensure they aren't the same)
        origin = random.choice(prefectures)
        destination = random.choice([p for p in prefectures if p != origin])
        
        # Add Seasonality (Heavy in Dec/March)
        # Japan Fiscal Year End is March, Oseibo is Dec.
        weight_multiplier = 1.0
        if current_date.month in [12, 3]: 
            weight_multiplier = 1.5
            
        weight = round(random.uniform(1.0, 25.0) * weight_multiplier, 1)
        
        data.append([
            current_date.strftime('%Y-%m-%d'),
            origin,
            destination,
            random.choice(vehicle_types),
            weight,
            random.choice(statuses)
        ])
        
    df = pd.DataFrame(data, columns=[
        "Date", "Origin_Prefecture", "Destination_Prefecture", 
        "Vehicle_Type", "Cargo_Weight_Tons", "Status"
    ])
    
    return df

df_large = generate_logi_data(10000)
df_large.to_csv('data/synthetic_logistics_data.csv', index=False, encoding='utf-8')

print("✅ Data Generation Complete: 10,000 rows created.")
print(df_large.head())
```
